```python
df.dropna(inplace=True)
```

```python
from textblob import TextBlob
```

```python
TextBlob("This video is amazing").sentiment.polarity
```

```python
sample_df = df.head(1000).copy()
```

```python
polarity = []

for comment in sample_df["comment_text"]:
    try:
        polarity.append(TextBlob(str(comment)).sentiment.polarity)
    except:
        polarity.append(0)

sample_df["polarity"] = polarity
```

```python
sample_df[["comment_text", "polarity"]].head(10)
```

## Sentiment Classification

```python
def sentiment(score):
    if score > 0.3:
        return "Positive"
    elif score < -0.3:
        return "Negative"
    else:
        return "Neutral"

sample_df["sentiment"] = sample_df["polarity"].apply(sentiment)
```

```python
sample_df[["comment_text", "polarity", "sentiment"]].head(10)
```

```python
sample_df["sentiment"].value_counts()
```

## Sentiment Visualization

```python
sample_df["sentiment"].value_counts().plot(kind="bar")

plt.title("Sentiment Analysis")
plt.xlabel("Sentiment")
plt.ylabel("Number of Comments")
plt.xticks(rotation=0)
plt.show()
```

```python
comments_positive = sample_df[sample_df["sentiment"] == "Positive"]

comments_negative = sample_df[sample_df["sentiment"] == "Negative"]
```

```python
all_emoji_list = []

for i in df["comment_text"].dropna():
    for j in str(i):
        if j in emoji.EMOJI_DATA:
            all_emoji_list.append(j)
```

```python
all_emoji_list[0:10]
```

```python
from collections import Counter
```

```python
Counter(all_emoji_list).most_common(10)
```

```python
emojis = [Counter(all_emoji_list).most_common(10)[i][0]
    for i in range(10)]

emojis
```

```python
freqs = [Counter(all_emoji_list).most_common(10)[i][1]
    for i in range(10)]

freqs
```

## Emoji Visualization

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.bar(emojis, freqs)

plt.title("Top 10 Most Used Emojis")
plt.xlabel("Emojis")
plt.ylabel("Frequency")
plt.show()
```
